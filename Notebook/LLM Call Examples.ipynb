```python
import boto3
import json

# Initialize the Bedrock runtime client
# Replace 'your-aws-region' with your actual AWS region (e.g., 'us-west-2')
bedrock_runtime_client = boto3.client(
    service_name='bedrock-runtime',
    region_name='us-east-2'  # Replace with your region
)

# Specify the model ID for Claude Sonnet 4
model_id = "us.anthropic.claude-sonnet-4-20250514-v1:0"

# Define the prompt for the model
user_message = "Tell me a joke for a 7 year old"

# Construct the message payload for the Converse API
# The Converse API provides a consistent way to interact with various Bedrock models.
messages = [
    {
        "role": "user",
        "content": [{"text": user_message}]
    }
]

# Configure inference parameters (optional)
inference_config = {
    "maxTokens": 100,  # Maximum number of tokens to generate
    "temperature": 0.7, # Controls randomness in output
    "topP": 0.9          # Nucleus sampling parameter
}

try:
    # Invoke the model using the Converse API
    response = bedrock_runtime_client.converse(
        modelId=model_id,
        messages=messages,
        inferenceConfig=inference_config
    )

    # Extract the response text
    response_text = response['output']['message']['content'][0]['text']
    print("Claude Sonnet 4 Response:")
    print(response_text)

except Exception as e:
    print(f"Error invoking Claude Sonnet 4: {e}")
```

```python
import boto3
import json

# Replace 'your-aws-region' with the region where you enabled the model (e.g., 'us-west-2')
region_name = 'us-east-2'  # Replace with your region
model_id = 'us.anthropic.claude-sonnet-4-20250514-v1:0'

# Initialize the Bedrock Runtime client
bedrock_runtime = boto3.client(
    service_name='bedrock-runtime',
    region_name=region_name
)

user_prompt = "What are the three most popular tourist attractions in Paris?"

# Construct the body according to Anthropic's Messages API
request_body = {
    "anthropic_version": "bedrock-2023-05-31", # Required for the Messages API
    "max_tokens": 2048,
    "temperature": 0.5,
    "messages": [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": user_prompt}
            ]
        }
    ]
}

try:
    response = bedrock_runtime.invoke_model(
        modelId=model_id,
        body=json.dumps(request_body),
        contentType='application/json',
        accept='application/json'
    )

    # The response is a stream of bytes, which needs to be read and parsed
    response_body = json.loads(response.get('body').read())
    response_text = response_body['content'][0]['text']

    print("Model Response:\n", response_text)

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
import boto3
import json

from botocore.exceptions import ClientError

# Create a Bedrock Runtime client in the AWS Region of your choice.
client = boto3.client("bedrock-runtime", region_name="us-east-2")

# Set the model ID, e.g., Claude 3 Haiku.
model_id = "us.anthropic.claude-sonnet-4-20250514-v1:0"

# Define the prompt for the model.
prompt = "Describe the purpose of a 'hello world' program in one line."

# Format the request payload using the model's native structure.
native_request = {
    "anthropic_version": "bedrock-2023-05-31",
    "max_tokens": 512,
    "temperature": 0.5,
    "messages": [
        {
            "role": "user",
            "content": [{"type": "text", "text": prompt}],
        }
    ],
}

# Convert the native request to JSON.
request = json.dumps(native_request)

try:
    # Invoke the model with the request.
    response = client.invoke_model(modelId=model_id, body=request)
    # Decode the response body.
    model_response = json.loads(response["body"].read())

    # Extract and print the response text.
    response_text = model_response["content"][0]["text"]
    print(response_text)

except (ClientError, Exception) as e:
    print(f"ERROR: Can't invoke '{model_id}'. Reason: {e}")
    exit(1)


```

```python
import boto3
import json
from pathlib import Path
from dotenv import load_dotenv
import os

# --- Configuration ---
# Boto3 will automatically use the API key if it is set in the 
# AWS_BEARER_TOKEN_BEDROCK environment variable.

MODEL_ID = "us.anthropic.claude-sonnet-4-20250514-v1:0"
REGION_NAME = "us-east-2"  # Replace with the region your API key is configured for

def invoke_bedrock_model(prompt):
    """
    Invokes a Bedrock model using the Boto3 client.
    Authentication is handled automatically by Boto3, which prioritizes 
    the AWS_BEARER_TOKEN_BEDROCK environment variable if set.
    """
    global_env_path = Path.home() / ".env"
    load_dotenv(global_env_path,override=True)
    bedrock_api_key = os.getenv("AWS_BEARER_TOKEN_BEDROCK")
    # Check if the API key is set for user clarity
    if bedrock_api_key is None:
        print("!!! WARNING: The AWS_BEARER_TOKEN_BEDROCK environment variable is not set.")
        print("Falling back to default AWS credential chain (e.g., IAM role or shared config).")
        print("To use the Bedrock API Key, please set this environment variable.")
    
    try:
        # 1. Initialize the Bedrock Runtime client
        # Boto3 handles the authentication chain internally, picking up the 
        # API key from the environment variable if available.
        client = boto3.client(
            service_name="bedrock-runtime",
            region_name=REGION_NAME
        )

        # 2. Define the request body for Claude (Anthropic)
        # Note: Model families (Claude, Mistral, Llama) use different JSON formats.
        request_body = {
            "anthropic_version": "bedrock-2023-05-31",
            "max_tokens": 512,
            "messages": [
                {
                    "role": "user",
                    "content": [{"type": "text", "text": prompt}]
                }
            ]
        }
        
        # 3. Invoke the model
        response = client.invoke_model(
            body=json.dumps(request_body),
            modelId=MODEL_ID,
            contentType="application/json",
            accept="application/json"
        )
        
        # 4. Parse the response
        response_body = response['body'].read().decode('utf-8')
        response_json = json.loads(response_body)
        
        # Extract the model's text output
        model_output = response_json['content'][0]['text']
        
        return model_output
    
    except Exception as e:
        print(f"An error occurred during API invocation: {e}")
        return None

# --- Main Execution ---
if __name__ == "__main__":
    test_prompt = "Explain in one sentence why the Bedrock API key is easier for quick start integrations than IAM roles."
    
    print(f"Sending prompt to {MODEL_ID}...")
    print("-" * 30)

    result = invoke_bedrock_model(test_prompt)
    
    if result:
        print("✅ Model Response:")
        print(result)

    print("-" * 30)
    print("Execution complete.")

# EOF Marker is omitted for the final output as it's handled by the system.
```

```python
import os
import boto3
from botocore.exceptions import ClientError, NoCredentialsError

# --- Authentication Check ---

def check_aws_auth_method():
    # 1. Check for the Bedrock API Key environment variable first
    if 'AWS_BEARER_TOKEN_BEDROCK' in os.environ:
        print("✅ Bedrock API Key found:")
        print("   The application is configured to use the AWS_BEARER_TOKEN_BEDROCK environment variable.")
        print("   This takes priority for Bedrock calls over standard IAM credentials.")
        return

    # 2. If no API Key is found, try the standard IAM/Access Key mechanism
    try:
        # Get an STS client. STS calls always require standard IAM credentials (SigV4 signing).
        sts_client = boto3.client('sts')
        
        # This call will fail if only a Bedrock API Key (Bearer Token) is available
        response = sts_client.get_caller_identity()
        
        # If successful, it's using an IAM Role or Access Key
        print("✅ Standard AWS Credentials (IAM) found:")
        print(f"   Principal ARN: {response.get('Arn')}")
        
        # Check if it's an assumed role (like an EC2 instance profile or Lambda role)
        if ':assumed-role/' in response.get('Arn', ''):
            print("   Source: Assumed IAM Role (Instance Profile or Lambda Role)")
        else:
            print("   Source: IAM User Access Keys (Explicitly set or from local files)")
            
    except NoCredentialsError:
        print("❌ No Standard AWS credentials found (neither explicit keys nor an IAM Role).")
        print("   If Bedrock calls succeed, the Bedrock API Key is likely being passed directly in the client configuration or via an environment variable not covered here.")
        
    except ClientError as e:
        # This catches specific access denial issues, which means credentials were found but unauthorized
        if e.response['Error']['Code'] == 'InvalidClientTokenId':
             print("❌ Invalid Access Key ID found. Credentials exist but are invalid.")
        else:
             print(f"⚠️ An unexpected ClientError occurred: {e}. Credentials might be present but lack STS permissions.")


if __name__ == '__main__':
    check_aws_auth_method()
```
